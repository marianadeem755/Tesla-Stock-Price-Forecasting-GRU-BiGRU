### **`Candlestick chart:`**
* is used to determine possible price movement based on past patterns. 
* It is useful when trading as they show four price points (open, close, high, and low) throughout the period.
* A "candlestick pattern" is a movement in prices shown graphically on a candlestick chart. 
* The separation  on the chart, is said to be caused by an exhaustion gap and the subsequent move in the opposite direction occurs as a result of a breakaway gap.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.lines import Line2D
from matplotlib.patches import Rectangle

# prepare data
data = df.copy()
if 'Date' not in data.columns:
    data = data.reset_index()
data['Date'] = pd.to_datetime(data['Date'], errors='coerce')
price_columns = ['Open', 'High', 'Low', 'Close']
for col in price_columns:
    data[col] = pd.to_numeric(data[col], errors='coerce')
data = data.dropna(subset=['Date'] + price_columns).sort_values('Date').drop_duplicates(subset='Date').reset_index(drop=True)
data['Date_Num'] = mdates.date2num(data['Date'])

# important: define latest from the complete candlestick dataset
latest = data.iloc[-1]
highest = data.loc[data['High'].idxmax()]
lowest = data.loc[data['Low'].idxmin()]
up_color = '#ff8c00'
down_color = '#FF1493'
trend_color = '#6A0572'
navy = '#3B174D'
background = '#FFFCF7'
grid_color = '#EAE4EF'

# create candlestick chart
fig, ax = plt.subplots(figsize=(18, 8))
dates = data['Date_Num'].to_numpy()
if len(dates) > 1:
    candle_width = min(pd.Series(dates).diff().dropna().median() * 0.7, 0.7)
else:
    candle_width = 0.6
for _, row in data.iterrows():
    date_num = row['Date_Num']
    open_price = row['Open']
    close_price = row['Close']
    high_price = row['High']
    low_price = row['Low']
    color = up_color if close_price >= open_price else down_color
    ax.plot([date_num, date_num], [low_price, high_price], color=color, linewidth=1.0, zorder=2)
    body_bottom = min(open_price, close_price)
    body_height = abs(close_price - open_price)
    if body_height == 0:
        body_height = max((high_price - low_price) * 0.005, 0.01)
    ax.add_patch(Rectangle((date_num - candle_width / 2, body_bottom), candle_width, body_height, facecolor=color, edgecolor=color, linewidth=0.6, zorder=3))
ax.plot(data['Date_Num'], data['Close'], color=trend_color, linewidth=1.0, alpha=0.55, label='Closing Price Trend', zorder=1)

# highlight highest high and lowest low
ax.scatter(highest['Date_Num'], highest['High'], color=up_color, edgecolor='white', s=90, zorder=5)
ax.annotate(f"Highest: ${highest['High']:,.2f}", (highest['Date_Num'], highest['High']), xytext=(0, 12), textcoords='offset points', ha='center', fontsize=9, fontweight='bold', color=up_color)
ax.scatter(lowest['Date_Num'], lowest['Low'], color=down_color, edgecolor='white', s=90, zorder=5)
ax.annotate(f"Lowest: ${lowest['Low']:,.2f}", (lowest['Date_Num'], lowest['Low']), xytext=(0, -20), textcoords='offset points', ha='center', fontsize=9, fontweight='bold', color=down_color)

# titles, axes and legend
ax.set_title('Tesla Stock Market Price: Candlestick Chart', fontsize=20, fontweight='bold', color=navy, pad=20)
ax.set_xlabel('Date', fontsize=12, fontweight='bold')
ax.set_ylabel('Tesla Stock Price ($)', fontsize=12, fontweight='bold')
ax.xaxis_date()
ax.xaxis.set_major_locator(mdates.AutoDateLocator())
ax.xaxis.set_major_formatter(mdates.ConciseDateFormatter(ax.xaxis.get_major_locator()))
ax.grid(axis='y', color=grid_color, linestyle='--', alpha=0.8)
ax.set_axisbelow(True)
ax.set_facecolor(background)
fig.patch.set_facecolor('white')
legend_items = [Line2D([0], [0], color=up_color, marker='s', linestyle='None', markersize=9, label='Increasing Price'), Line2D([0], [0], color=down_color, marker='s', linestyle='None', markersize=9, label='Decreasing Price'), Line2D([0], [0], color=trend_color, linewidth=1.5, label='Closing Price Trend')]
ax.legend(handles=legend_items, loc='upper left', bbox_to_anchor=(1.01, 1), frameon=True, facecolor='white', edgecolor='#AB83A1', fontsize=10, title='Chart Legend', title_fontsize=11)

# latest trading summary
daily_change = latest['Close'] - latest['Open']
daily_change_pct = (daily_change / latest['Open'] * 100) if latest['Open'] != 0 else np.nan
summary = f"Latest Close: ${latest['Close']:,.2f}   |   Open: ${latest['Open']:,.2f}   |   Daily Change: {daily_change_pct:+.2f}%   |   Range: ${latest['Low']:,.2f}–${latest['High']:,.2f}"
fig.text(0.43, 0.025, summary, ha='center', fontsize=10, fontweight='bold', color=navy)
fig.subplots_adjust(left=0.08, right=0.80, top=0.88, bottom=0.13)
plt.show()

# print key insights
print("Tesla Candlestick Insights")
print("============================================================================")
print(f"Latest date : {latest['Date'].strftime('%d %B %Y')}")
print(f"Latest open : ${latest['Open']:,.2f}")
print(f"Latest close : ${latest['Close']:,.2f}")
print(f"Latest high : ${latest['High']:,.2f}")
print(f"Latest low : ${latest['Low']:,.2f}")
print(f"Daily price change : {daily_change:+,.2f}")
print(f"Daily change (%) : {daily_change_pct:+.2f}%")
print(f"Highest high : ${highest['High']:,.2f} on {highest['Date'].strftime('%d %b %Y')}")
print(f"Lowest low : ${lowest['Low']:,.2f} on {lowest['Date'].strftime('%d %b %Y')}")
```

### `Fan Chart:`
* A fan chart is a chart that joins a simple line chart for observed past data with ranges for possible values of future data.
*  The historical data and possible future data are joined with a line showing a central estimate or most likely value for the future outcomes.
*  The Fan Chart represents the forecasting distribution of a variable based on the information available at present
*  it depicts the whole marginal forecast distribution
*  The upper Bound is created by adding noise to the forecasted prices
*  The lower bound is created by subtracting noise from the forecasted prices

```python
# Prepare historical data
data = df.copy()
if 'Date' not in data.columns:
    data = data.reset_index()
data['Date'] = pd.to_datetime(data['Date'])
price_col = 'Adj Close' if 'Adj Close' in data.columns else 'Close'
data = data[['Date', price_col]].dropna().sort_values('Date')
data = data.drop_duplicates(subset='Date').reset_index(drop=True)
dates = data['Date']
prices = data[price_col].astype(float)
historical_avg = prices.mean()
last_price = prices.iloc[-1]

# estimate daily returns and historical volatility
log_returns = np.log(prices / prices.shift(1)).dropna()
daily_drift = log_returns.mean()
daily_volatility = log_returns.std()
forecast_days = 60
future_dates = pd.bdate_range(start=dates.iloc[-1] + pd.Timedelta(days=1), periods=forecast_days)
steps = np.arange(1, forecast_days + 1)
forecast_prices = last_price * np.exp(daily_drift * steps)
upper_bound = forecast_prices * np.exp(1.96 * daily_volatility * np.sqrt(steps))
lower_bound = forecast_prices * np.exp(-1.96 * daily_volatility * np.sqrt(steps))

# calculate historical insights
highest_idx = prices.idxmax()
lowest_idx = prices.idxmin()
period_change = (prices.iloc[-1] / prices.iloc[0] - 1) * 100
# create the chart
fig, ax = plt.subplots(figsize=(17, 9))
ax.plot(dates, prices, color='#6A0572', linewidth=1.8, label='Historical Adjusted Closing Price', zorder=3)
ax.axhline(historical_avg, color='#ff8c00', linestyle='--', linewidth=2, label=f'Historical Average: ${historical_avg:,.2f}')
ax.fill_between(future_dates, lower_bound, upper_bound, color='#FF1493', alpha=0.18, label='Illustrative 95% Uncertainty Range')
ax.plot(future_dates, forecast_prices, color='#FF1493', linewidth=2.5, linestyle='--', label='Illustrative Forecast')
ax.axvline(dates.iloc[-1], color='#555555', linestyle=':', linewidth=1.8)
ax.text(dates.iloc[-1], 0.97, ' Forecast starts', transform=ax.get_xaxis_transform(), ha='right', va='top', fontsize=10, color='#6A0572', fontweight='bold')
ax.scatter(dates.iloc[highest_idx], prices.iloc[highest_idx], color='#ff8c00', edgecolor='white', s=100, zorder=5, label=f'Historical High: ${prices.iloc[highest_idx]:,.2f}')
ax.scatter(dates.iloc[lowest_idx], prices.iloc[lowest_idx], color='#20ABAC', edgecolor='white', s=100, zorder=5, label=f'Historical Low: ${prices.iloc[lowest_idx]:,.2f}')
ax.annotate(f'Last Price: ${last_price:,.2f}', xy=(dates.iloc[-1], last_price), xytext=(12, 12), textcoords='offset points', fontsize=10, fontweight='bold', color='#6A0572', arrowprops=dict(arrowstyle='->', color='#6A0572'))
ax.annotate(f'Forecast: ${forecast_prices[-1]:,.2f}', xy=(future_dates[-1], forecast_prices[-1]), xytext=(-10, 12), textcoords='offset points', ha='right', fontsize=10, fontweight='bold', color='#FF1493')
ax.set_title('Tesla Stock: Historical Performance & Illustrative Forecast', fontsize=20, fontweight='bold', color='#3B174D', pad=20)
ax.set_xlabel('Date', fontsize=12, fontweight='bold')
ax.set_ylabel('Adjusted Stock Price ($)', fontsize=12, fontweight='bold')
ax.xaxis.set_major_locator(mdates.AutoDateLocator())
ax.xaxis.set_major_formatter(mdates.ConciseDateFormatter(ax.xaxis.get_major_locator()))
ax.grid(True, linestyle='--', alpha=0.22)
ax.set_facecolor('#FFFCF7')
summary = f'Historical change: {period_change:+.2f}%   |   Daily volatility: {daily_volatility * 100:.2f}%   |   Forecast horizon: {forecast_days} trading days'
fig.text(0.5, 0.015, summary, ha='center', fontsize=11, color='#3B174D', fontweight='bold')
ax.legend(loc='upper left', fontsize=9, frameon=True, framealpha=0.95, ncol=2)
fig.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()

print("Tesla Stock Analysis")
print("===========================================================================")
print(f"Latest historical price : ${last_price:,.2f}")
print(f"Historical average : ${historical_avg:,.2f}")
print(f"Historical period change: {period_change:+.2f}%")
print(f"Historical highest price: ${prices.max():,.2f} ({dates.iloc[highest_idx].date()})")
print(f"Historical lowest price : ${prices.min():,.2f} ({dates.iloc[lowest_idx].date()})")
print(f"Daily volatility : {daily_volatility * 100:.2f}%")
print(f"Forecast endpoint : {future_dates[-1].date()}")
print(f"Illustrative forecast : ${forecast_prices[-1]:,.2f}")
print(f"Forecast range : ${lower_bound[-1]:,.2f} to ${upper_bound[-1]:,.2f}")
```

* This Fan chart provides the visualization of Historical stock prices along with upper and lower bound 
* The upper and lower bound represents the uncertainity associated with forecast as the upper and lower bound depicts the possible predictions and outcomes
* The random deviations are aadded or subtracted from the median forecasted stock prices in order to create uper and lower bound
* The shaded region between the upper and lower bound in this plot represents the range of possible outcomes.Hence as the shaded region becomes wider the uncertainity caueses to be increased

### **`Moving Average (MA) Chart:`**
* A moving average (MA) chart is used to track the stock price movements. 
* It plots average prices over a defined period of time
* A moving average (MA) particularly used to help smooth out price data by creating a constantly updated average price.
* A rising moving average indicates that the stock prices is in an uptrend, while a declining moving average indicates a downtrend

### **`Decomposition:`**
Decomposition in time series analysis is a method used to separate a time series into three distinct components: trend, seasonality, and residual (or irregular) component
#### **`Trend:`**
* Long-term increase or decrease in the data. 
* The trend can be any function, such as linear or exponential, and can change direction over time. 
#### **`Seasonality:`**
* Seasonality is the Repeating cycle in the series with fixed frequencies.
#### **`Residual:`**
* The residual component is obtained after removing the trend and seasonal components from the time series.
* In time series analysis, residuals, r, are the difference between the fitted values and the actual values

### `BiGRU (Bidirectional GRU):`
* A Bidirectional GRU, or BiGRU, is a sequence processing model that consists of two GRUs.
* one taking the input in a forward direction, and the other in a backwards direction.
*  It is a bidirectional recurrent neural network with only the input and forget gates.

```python
# Set random seeds for reproducibility
random.seed(42)
np.random.seed(42)
tf.keras.utils.set_random_seed(42)
tf.config.experimental.enable_op_determinism()
close_data = df['Close'].astype(float).values
if len(close_data) <= 100:
    raise ValueError("Not enough observations for the selected time-series window.")
if not np.isfinite(close_data).all():
    raise ValueError("Close prices contain missing or infinite values.")
n = len(close_data)
train_end = int(n * 0.70)
val_end = int(n * 0.85)
train_raw = close_data[:train_end]
val_raw = close_data[train_end:val_end]
test_raw = close_data[val_end:]

# Fit scaler on training data only
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_raw.reshape(-1, 1)).flatten()
val_scaled = scaler.transform(val_raw.reshape(-1, 1)).flatten()
test_scaled = scaler.transform(test_raw.reshape(-1, 1)).flatten()

# Create sequences using historical observations
def create_sequences(data, window_size):
    X, y = [], []

    for i in range(window_size, len(data)):
        X.append(data[i - window_size:i])
        y.append(data[i])

    return np.array(X), np.array(y)

window_size = 30

# Training sequences
X_train, y_train = create_sequences(train_scaled, window_size)

# Validation sequences include preceding training observations as context
val_with_context = np.concatenate([train_scaled[-window_size:], val_scaled])
X_val, y_val = create_sequences(val_with_context, window_size)

# Testing sequences include preceding validation observations as context
test_context = np.concatenate([train_scaled, val_scaled])
test_with_context = np.concatenate([test_context[-window_size:], test_scaled])
X_test, y_test = create_sequences(test_with_context, window_size)

# Reshape sequences for GRU input
X_train = X_train.reshape(-1, window_size, 1)
X_val = X_val.reshape(-1, window_size, 1)
X_test = X_test.reshape(-1, window_size, 1)

# Define the BiGRU model
model = Sequential([
    tf.keras.Input(shape=(window_size, 1)),
    Bidirectional(GRU(32, return_sequences=True)),
    Dropout(0.2),
    Bidirectional(GRU(16)),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(optimizer=Adam(learning_rate=0.001),loss='mse',metrics=['mae'])
# Stop training when validation performance stops improving
early_stopping = EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=4,
    min_lr=0.00001,
    verbose=1
)

# Train the model
history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=32,
    shuffle=False,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)

# Predict normalized closing prices
y_pred = model.predict(X_test, verbose=0).flatten()

# Convert predictions and actual values to original prices
y_test_original = scaler.inverse_transform(y_test.reshape(-1, 1)).flatten()
y_pred_original = scaler.inverse_transform(y_pred.reshape(-1, 1)).flatten()

# Evaluate BiGRU predictions
test_mse = mean_squared_error(y_test_original, y_pred_original)
test_rmse = np.sqrt(test_mse)
test_mae = mean_absolute_error(y_test_original, y_pred_original)
test_r2 = r2_score(y_test_original, y_pred_original)
baseline_pred = test_context[-1:]  

baseline_scaled = test_with_context[window_size - 1:-1]
baseline_original = scaler.inverse_transform(
    baseline_scaled.reshape(-1, 1)
).flatten()

baseline_mse = mean_squared_error(y_test_original, baseline_original)
baseline_rmse = np.sqrt(baseline_mse)
baseline_mae = mean_absolute_error(y_test_original, baseline_original)
baseline_r2 = r2_score(y_test_original, baseline_original)
print(f"MAE:  {baseline_mae:.4f}")
print(f"RMSE: {baseline_rmse:.4f}")
print(f"MSE:  {baseline_mse:.4f}")
print(f"R2 Score: {baseline_r2:.4f}")
```
